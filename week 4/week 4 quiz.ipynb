### 1. A multiverse analysis is designed to assess what type of uncertainty?


- aleatory uncertainty
- parameter uncertainty
- predictive uncertainty
- [x] large-world uncertainty
- small-world uncertainty

**Answer: large-world uncertainty**

A **multiverse analysis** assesses uncertainty that comes from **researcher choices** and **modeling/analysis decisions**, such as which variables to include, how to clean the data, which statistical model to use, or which exclusion criteria to apply.

That makes it a form of **large-world uncertainty**, because the uncertainty is about the broader analytic process and possible reasonable worlds of analysis, not just randomness or parameter estimates within one fixed model.

### 5. In Bayesian data analysis, which distribution links the data space and parameter space.

- Prior distribution
- [x] Likelihood distribution
- Posterior distribution
- Normal distribution

**Answer: Likelihood distribution**

In Bayesian data analysis, the **likelihood distribution** links the **data space** and the **parameter space** because it expresses how likely the observed data are under different possible parameter values.

Formula-wise:

$
P(\theta \mid y) \propto P(y \mid \theta)P(\theta)
$

Where:

* $ (P(\theta)) $ = prior distribution
* $ (P(y \mid \theta)) $ = **likelihood distribution**
* $ (P(\theta \mid y)) $ = posterior distribution


### 6. According to course lectures, one advantage of a credible interval over a confidence interval is:


- The credible interval is generally more accurate than the confidence interval.
- The credible interval is generally larger than the confidence interval.
- The credible interval can be calculated with fewer degrees of freedom (DOF) than the confidence interval.
- [x] The credible has a more intuitive interpretation than the confidence interval.

A credible interval can be interpreted as: given the observed data and prior assumptions, there is a specified probability that the parameter falls within the interval.

For example, a 95% credible interval means there is a 95% posterior probability that the parameter lies in that interval.

A 95% confidence interval does not technically mean there is a 95% probability that the true parameter is inside this specific interval; it means that, over repeated sampling, the procedure would capture the true parameter 95% of the time.

### 8. In Bayesian data analysis, the likelihood function represents:


- [x] The probability of observing the data given the parameters
- The probability of observing the parameters given the data
- The posterior distribution of the parameters
- The credible intervals for Bayesian predictions

**Answer: The probability of observing the data given the parameters**

In Bayesian analysis, the **likelihood function** is:

$
P(\text{data} \mid \text{parameters})
$

It tells us how compatible different parameter values are with the observed data. The posterior reverses the conditioning:

$
P(\text{parameters} \mid \text{data})
$

### 10. As discussed in the lectures, which of the following is a benefit of using Bayesian data analysis?


- It provides exact and deterministic results for all statistical models.
- It allows for straightforward interpretation of p-values.
- [x] It incorporates prior information and updates beliefs based on observed data.
- It requires fewer assumptions compared to classical frequentist analysis.

**Answer: It incorporates prior information and updates beliefs based on observed data.**

Bayesian data analysis is useful because it combines:

$
\text{Prior beliefs} + \text{Observed data} \rightarrow \text{Updated posterior beliefs}
$

So it allows analysts to formally include prior evidence or expert knowledge, then revise those beliefs after seeing the data.

- It requires the assumption of a specific parametric distribution for the data.

Bayesian analysis often requires computational methods like MCMC or other sampling/approximation techniques, especially for complex models. These methods can become slow or resource-intensive with large datasets or many parameters.
