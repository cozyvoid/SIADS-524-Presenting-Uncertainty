# It's a Large World, After All

---

*"Model uncertainty is a central challenge of modern science. Researchers do not know what model actually generated their data, and they cannot be sure if the statistical method they use is best. Moreover, researchers can explore innumerable analytical variations in their search for results, but readers can only see the handful of estimates curated for publication. These researcher degrees of freedom represent a crisis in science today: are empirical conclusions driven mostly by the data, or mostly by the analytical assumptions made by researchers?"*
> Cristobal Young

- Young is describing the problem of large world uncertainty
- To the extent to which formal education and training programs provide guidance in how to model and represent uncertainty.
- They tend to almost exclusively focus in on the small world of uncertainty.

### Large world uncertainty







1. small world uncertainty rests on large world uncertainty
2. large world itself is an important source of uncertainty

 - large world uncertainty is really difficult to completely quantify and that there's no formula that we can easily employ like we do in small world uncertainty
 - multiverse analysis involves performing a set of analyses across a range of reasonable design choices, impacting the data and the model.
 - precursor to multiverse analysis was called the outlier analysis, which considers the effect of potential outliers by showing the results of an analyses with or without the outliers included.

### Multiverse Analysis





> a multiverse analysis does not consider all aspects of large world uncertainty

> It does not directly address the underlying assumptions of our modeling choices, nor does it really look at the data generation process



- In multiverse analysis, we define the different reasonable choices we can make in modeling, including data processing, transformation, variable inclusion, model type.
- And we model all of those options, creating sometimes hundreds or thousands of models, perhaps even more.
- And then we share the results of all the models in some sort of visualization. And in doing so, we give a broad picture as to the nature of the underlying relationships in the data.
- In multiverse modeling, it is a model many share all approach to data science.



> Multiverse modeling is for describing large world uncertainty and understanding the generalizability of our findings.

### Replication Crisis



- collect data and analyze it using the same methods described in the original research and see if they can replicate the results
- One of the first significant attempts at a reproduction study is the Reproducibility Project in psychology, which replicated 100 original studies in psychology and found that 64% of the 97 studies failed to replicate
- On screen, each red square represents a failed study
- A follow up study in Nature magazine in 2018 found that 38% of the 21 studies published in Nature between 2010 and 2015 failed to replicate and the effect size of these replications are only 50% of the original effect size
- These studies describe what some call a replication crisis in science
- This replication crisis is really a problem of large world uncertainty

#### *How certain am I of the findings in a multiverse of possibilities?*





- Multiverse modeling is not about finding the single best model
- Rather than asking which model is best, Young encouraged researchers to ask which models deserve consideration
- Multiverse analysis simply helps us understand how robust and stable the findings of that one model are in the broader set of choices that we could have made

### How to use and not use multiverse analysis



### Multiverse Process



- Large world uncertainty is very important but difficult to model.
- One option to model some aspects of the large world is multiverse analysis.
- Multiverse analysis gives us a framework to understand how robust our overall finding is.
- There are aspects to large world uncertainty that multiverse analysis does not directly consider, such as the underlying assumptions of modeling choices and the data generation process.
