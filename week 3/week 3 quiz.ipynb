### 8. Which of the following is an example of a deterministic construal error?

- A user misreading a 95% confidence interval as a 90% confidence interval.
- [x] A user misreads the predictive interval around a predicted temperature value as the high and low temperature for that day.
- A user misstates a percentile as a percentage, as in a student scored in the 87th percentile on a MADS exam and the user misinterprets that to mean that the student got 87% of the questions correct.
- The user reads the relationship depicted in a graph as causal, when only correlation can be implied. For example, a user sees a graph that shows a positive relationship between income level and exercise and falsely assumes that the more people exercise the more their income will increase.
- A user misinterprets a 95% confidence interval to mean that there is a 95% chance that the true population parameter falls in the range of the confidence interval.

**A user misreads the predictive interval around a predicted temperature value as the high and low temperature for that day.**

*That is a deterministic construal error because the user is treating an uncertainty interval as if it were a deterministic forecast of actual values, rather than a range of plausible prediction uncertainty.*

### 9. What is the purpose of the 'finite sample adjustment' when calculating the quantile threshold from non-conformity scores?

- To force non-conformity scores to (generally) follow the normal distribution. 
- [x] To select a slightly more conservative threshold to ensure coverage guarantees are met even with a few calibration points.
- To make the prediction intervals adaptive and dependent on the features of each data point.
- To correct for a bias on the t-test of significance of the populations parameter for the prediction interval. 
- To make computations faster on larger (and often multi-dimensional) calibration sets.

**To select a slightly more conservative threshold to ensure coverage guarantees are met even with a few calibration points.**

*The finite sample adjustment is used in conformal prediction because the calibration set is finite, so the quantile cutoff must be chosen carefully to maintain the intended coverage level. It is not about making scores Normal, adapting intervals by features, correcting a t-test bias, or speeding computation.*
