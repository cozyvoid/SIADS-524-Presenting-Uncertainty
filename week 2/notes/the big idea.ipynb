# The Big Idea

---         

#### Data science is about generating an estimate of something we cannot be certain about.





## *point estimate*



#### The problem with uncertainty is that we often set it aside ot focus on *point estimate* of what we are tring to study (e.g., a statistic, a coefficient, a prediction, a difference, etc.) that we can't know for certain. 

##### It's important that we communicate it but without uncertainty it lacks context. 

### The popularity of point estimates

- sample statistics (i.e. mean hours of sleep of adults in Greece)
- models (i.e. regression models that relates to size of a Spanish home is square meters against its price in euros)
- survey results (44.3% of data scientist use SQL on a rgular basis )








#### The point is that in much of statistical practice, we produce point estiamtes for our output. The point estimate or single value is what users often expect. The question is; however, how much confidence do we have in this point estimate?

### The problem with point estimates



The point estimate may be the most likely value but we would be surprised if the true value differed from our point estimate in some meaningful way. The challenge with the previous examples discussed is that they **ignore uncertainty**. 

### What are our options when dealing with uncertainty?
#### Uncertainty is everywhere.

Uncertainty is fundamental to analytical problems - both small world and large world. What can we do?

- Not represent it - *ignore it*
- Gather more data to sometimes make our best estimates more certain - *minimize it* (impractical and not cost effective)
- Estimate it - *model it*

### REVIEW: Three ways to calculate the standard error 



We can use multiple samples from a population and we calculate the same statistic for each sample and produce a sampling distribution. From the sampling distribution we need the standard deviation, which, in this case, equivalent to the standard error. 

The challenges with this option is that we often don't have multiple samples from which to draw which leads us to the second method.



Depending on the statistic we are using we may be able to use a simple formula to estimate it analytically. Some statistics like mean and median, have a specific formula from which we can estimate the standard error from a single sample. 



Taking a single sample a resampling it thousands of times to create thousands of synthetic samples. From each we can calculate the statistic of interest and form a sampling distribution. We then can calcualte the standard deviation of that sampling distribution to find the standard error. 

### Let's see how these strategies can play out with an example...

Suppose I am studying the salaries of data scientists in Germany. While I am interested in the full population, I don't have access to it. So I draw a sample. 

I sample from data scientists in Germany, and from that sample I calculate several statistics, such as mean, standard deviation, etc. These two statistics describe be the sample and if they its randomly drawn and of sufficient size they can be used as a reasonable proxy for my populatin's mean and standard deviation. 

How well does that sample mean estimate the population mean?

To estimate this, we need to find the standard error of the mean (the standard deviation of the distribution of sample means)

### (3) Ways that we can get the standard error statistic. 

#### 1. Repeat sampling



Collect multiple samples and from each of those samples we would calculate a mean, compile those means up to make a sampling distribution of the means, adn read the standard error (uncertainty) from that sampling distribution. 

#### 2. Bootstrapping (synthetic option)



Where we take the one sample that we do have of data scientists in Germany and we resample it thousands of times. We draw the sample thousands of times, and ech time we draw the sample we calculate the mean. We pile up those means in a sampling distribution of the means, a nd we can read the uncertainty directly from that distributionl.

#### 3. Calculate analytically.



We need a formula to calculat the formula analytically off of a single sample and these many statistics that we don't have such a formula. In this case the mean has such a formula, the standard error = the standard deviation of the sample / the square root of the sample size of the sample. Once we create that calculation and the value from it, we can use that for our standard error and produce any number of uncertainty calculations and visualizations.
